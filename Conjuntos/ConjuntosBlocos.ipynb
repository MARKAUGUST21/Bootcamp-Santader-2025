```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

set3 = set1.intersection(set2) # ele pega os elementos comuns entre os dois conjuntos
print(set3)
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

set3 = set1.difference(set2) # ele pega os elementos que estão no set1 mas não estão no set2
print(set3)
```

```python
set = {1,2,3,5}
set2 = {6,7,8,9,10}

set3 = set.union(set2)
print(set3)
```
